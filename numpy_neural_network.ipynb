```python
class Dense:
  def __init__(self , x ):
    self.weight = np.random.rand(x.shape[1] , 32)* 0.01
    self.bias = np.zeros((1, 32))

  def forward(self , x):
    self.input = x
    return np.dot(x , self.weight) + self.bias

  def backward(self , dY , lr =0.01):
    dW = np.dot(self.input.T , dY)
    dB = np.sum(dY , axis=0 , keepdims=True)
    dX = np.dot(dY , self.weight.T)

    self.weight = self.weight - lr*dX
    self.bias = self.bias - lr*dB

    return dX
```

```python
import numpy as np

class SoftmaxCrossEntropy:
    def forward(self, logits, y_true):
        self.y_true = y_true
        shifted = logits - np.max(logits, axis=1, keepdims=True)
        exp_vals = np.exp(shifted)
        self.probs = exp_vals / np.sum(exp_vals, axis=1, keepdims=True)

        batch_size = logits.shape[0]
        correct_class_probs = self.probs[np.arange(batch_size) , y_true]
        loss = -np.log(correct_class_probs , 1e-9)
        return np.mean(loss)

    def backwrd(self , y_true):
        batch_size = self.probs.shape[0]
        dY = self.probs.copy()
        self.probs[np.arange(batch_size), y_true] -=1
        dY = dY / batch_size
        return dY
```
