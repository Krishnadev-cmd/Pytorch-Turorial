```python
import numpy as np
```

```python
class Dense:
  def __init__(self , x ):
    self.weight = np.random.rand(x.shape[1] , 32)* 0.01
    self.bias = np.zeros((1, 32))

  def forward(self , x):
    return np.dot(x , self.weight) + self.bias
```

```python
class Relu:
  def __init__(self , x):
    return max(x , 0)
```

```python
x = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

layer = Dense(x)
out = layer.forward(x)

print(out.shape)
```
