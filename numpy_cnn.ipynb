```python
import numpy as np
```

```python
class Conv2d:
  def __init__(self , output_channel , input_channel , kernel):
    self.output_channel = output_channel
    self.input_channel = input_channel
    self.K = kernel
    self.weights = np.random.rand(output_channel , input_channel , kernel , kernel)
    self.bias  = np.zeros((output_channel))

  def forward(self , input , stride , padding):
    self.input = input
    C , H , W = self.input.shape
    h_out = (H - self.K)//stride+1
    w_out = (W - self.K)//stride+1

    output = np.zeros((self.output_channel , h_out , w_out))

    for c in range(self.output_channel):
      for h in range(h_out):
        for w in range(w_out):
          region = self.input[: , h:h+self.K , w:w+self.K]
          output[c , h , w] = np.sum(self.weights[c] , region) + self.bias[c]
    return output
```

```python
class ReLU:
  def forward(self , x):
    self.input = x
    return np.max(self.input , 0)

  def backward(self , dZ):
    dZ[self.input <=0]=0
    return dZ
```

```python
class Dense:
  def __init__(self,output_channel , input_channel , kernel):
    self.weights = np.random.rand(output_channel , input_channel , kernel , kernel)
    self.bias  = np.zeros((output_channel))

  def forward(self , x):
    self.input = x
    return np.dot(self.weights , x) + self.bias

  def backward(self ,dZ):
    dW = np.dot(self.input.T , dZ)
    db = np.sum(dZ , axis=1 , keepdims=True)
    dx = np.dot(self.weights.T , dZ)
    return dx
```

```python
class MaxPool:
  def forward(self , input , stride):
    self.input = input
    C , H , W = self.input.shape
    h_out = (H - self.K)//stride+1
    w_out = (W - self.K)//stride+1
    output = np.zeros((C , h_out , w_out))

    for c in range(C):
      for h in range(h_out):
        for w in range(w_out):
          region = self.input[c , h:h+self.K , w:w+self.K]
          output[c ,h, w] = np.max(region)
    return output
```
