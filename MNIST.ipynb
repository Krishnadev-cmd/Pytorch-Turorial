```python
import numpy as np
```

```python
class Conv2d:
  def __init__(self , input_channel , output_channel , kernel , stride):
    self.input_channel = input_channel
    self.output_channel = output_channel
    self.kernel = kernel
    self.stride = stride

    self.weight=  np.random.randn(output_channel , input_channel , kernel , stride)
    self.bias = np.zeros(output_channel)

  def forward(self ,x):
    channel , height , width = x.shape
    k = self.kernel

    h_out = height-k+1
    w_out = width-k+1

    feature_maps = np.zeroes((self.output_channels , h_out , w_out))

    for channels in range(self.output_channel):
      for h in range(h_out):
        for w in range(w_out):
          region = x[: , h:h+k , w:w+k]
          feature_maps[channels, h, w] = np.sum(region * self.weights[channels]) + self.bias[channels]

    return feature_maps
```
